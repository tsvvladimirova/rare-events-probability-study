# When Does the Unlikely Become Likely?

**A Mathematical Investigation of Rare Events Using Probability, Python Simulations and Statistical Analysis**

## Research Question

How does the number of independent trials affect the probability of observing a rare event at least once?

## Project Overview

This project investigates how the probability of observing a rare event at least once changes as the number of independent trials increases.

Although the numerical examples use a single-trial event probability of $p=0.2$, the mathematical model is general and can be applied to events with different probabilities, including much rarer events, provided that the assumptions of the model are satisfied.

The investigation combines probability theory, mathematical modelling, Python simulations
and statistical analysis.

## Research Objectives

The objectives of this project are to:

1. Develop and explain a mathematical model for the probability of observing a rare event at least once.
2. Formulate research hypotheses before conducting experiments.
3. Implement simulations in Python.
4. Compare theoretical probabilities with simulated results.
5. Analyze the findings and discuss the assumptions and limitations of the model.

## Research Hypotheses

**H1 - Effect of the Number of Trials**

If the probability of a rare event in each independent trial remains constant and satisfies $0<p<1$, then increasing the number of trials will increase the probability of observing the event at least once.

**H2 - Theoretical and simulated probabilities**

The empirical probability estimated through repeated simulations will tend to approach the theoretical probability as the number of simulation repetitions increases.

**H3 - Random variation**

The results of individual simulations will vary due to randomness. Therefore, the empirical probability will not necessarily get closer to the theoretical probability after every additional simulation.

## Mathematical Background

### 1. Mathematical Foundations for H1: Effect of the Number of Trials

#### Probability and Complementary Events

Probability measures how likely an event is to occur. Its value ranges from 0 to 1, where 0 represents an impossible event and 1 represents a certain event.

For any event A, either A occurs or its complement, denoted by Aᶜ, occurs. These two outcomes are mutually exclusive and together cover all possible outcomes.

Therefore, the probability of the complement is:

$$
P(A^c) = 1 - P(A)
$$

This relationship follows the complement rule (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).


Consider the following case: the probability of event $A$ is 0.2, or 20%.

The probability that event $A$ does not occur is:

$$
P(A^c)=1-P(A)=1-0.2=0.8
$$

Therefore, the probability of the event not occurring is 0.8, or 80%.

This illustrates that the probabilities of an event and its complement sum to 1.

#### Independent Trials

Two trials are independent when the outcome of one trial does not change the probability of an event occurring in the other trial (Illowsky & Dean, 2023, [Section 3.2](https://openstax.org/books/introductory-statistics-2e/pages/3-2-independent-and-mutually-exclusive-events)).
    

In this project, we assume that the trials are independent and that the probability of the event, denoted by $p$, remains constant across all trials. Keeping $p$ constant allows us to investigate how changing the number of trials, denoted by $n$, affects the probability of observing the event at least once.

The probability that the event does not occur in a single trial is:

$$
1-p
$$

For example, if the probability of an event in a single trial is $p = 0.2$, then the probability that it does not occur is:

$$
1-p=1-0.2=0.8
$$

Therefore, the probability of the event not occurring in a single trial is 0.8, or 80%.

#### Probability of No Occurrence in Multiple Independent Trials

If the trials are independent, the probability that the event does not occur in either of two trials is:

$$
(1-p)(1-p)=(1-p)^2
$$

For $p=0.2$, this probability is:

$$
(1-0.2)^2=0.8^2=0.64
$$

Therefore, the probability that the event does not occur in either of the two trials is 0.64, or 64%.

#### Probability of at Least One Occurrence

The probability of observing an event at least once is the complement of the probability that the event does not occur in any trial.

Using the complement rule, we obtain:

$$
P(\text{at least one occurrence})=1-P(\text{no occurrence})
$$

For $n$ independent trials, the probability of no occurrence is $(1-p)^n$. Therefore, the probability of observing the event at least once is:

$$
\boxed{P(\text{at least one occurrence})=1-(1-p)^n}
$$

In this formula, $p$ represents the probability of the event in a single trial, and $n$ represents the number of independent trials.

The multiplication rule for independent events states that the probability of all events occurring is the product of their individual probabilities (Illowsky & Dean, 2023, [Section 3.3](https://openstax.org/books/introductory-statistics-2e/pages/3-3-two-basic-rules-of-probability)).


#### Numerical Examples for H1

To investigate how the number of independent trials affects the probability of observing an event at least once, we keep the probability of the event in a single trial constant at $p=0.2$, or 20%, and vary the number of trials, denoted by $n$.

Keeping $p$ constant allows us to examine the effect of the number of trials without changing the probability of the event in each individual trial.

Using the formula derived above:

$$
P(\text{at least one occurrence})=1-(1-p)^n
$$

**Case 1: One trial ($n=1$)**

$$
P=1-(1-0.2)^1=1-0.8=0.2=20\%
$$

**Case 2: Two trials ($n=2$)**

$$
P=1-(1-0.2)^2=1-0.64=0.36=36\%
$$

**Case 3: Three trials ($n=3$)**

$$
P=1-(1-0.2)^3=1-0.512=0.488=48.8\%
$$

**Case 4: Five trials ($n=5$)**

$$
P=1-(1-0.2)^5=1-0.32768=0.67232=67.232\%
$$

**Case 5: Ten trials ($n=10$)**

$$
P=1-(1-0.2)^{10}=1-0.1073741824=0.8926258176\approx89.26\%
$$

The results can be summarized as follows:

| Number of trials ($n$) | Probability of at least one occurrence |
|---:|---:|
| 1 | 20.00% |
| 2 | 36.00% |
| 3 | 48.80% |
| 5 | 67.23% |
| 10 | 89.26% |

**Observation**

The theoretical probability of observing the event at least once increases as the number of independent trials increases, even though the probability in each individual trial remains constant at 20%.

The increase is not linear: adding more trials does not produce the same increase in percentage points each time. The probability approaches 100% as the number of trials increases, provided that $p>0$ remains constant and the trials are independent.


### 2. Mathematical Foundations for H2: Theoretical and Empirical Probability

#### Theoretical Probability

The theoretical probability is the probability calculated using a mathematical model. In this project, it is calculated using the formula derived above:

$$
P=1-(1-p)^n
$$

This value provides a mathematical reference against which the simulation results can be compared.

#### Empirical Probability and Relative Frequency

The empirical probability is estimated from the results of repeated simulations. It is calculated as the proportion of simulations in which the event occurs at least once.

Let $N$ denote the total number of simulations and $k$ denote the number of simulations in which the event occurs at least once. The empirical probability is:

$$
\hat{P}=\frac{k}{N}
$$

Here, $\hat{P}$ represents the empirical estimate of the theoretical probability $P$.

For example, if the event occurs at least once in 354 out of 1,000 simulations, the empirical probability is:

$$
\hat{P}=\frac{354}{1000}=0.354=35.4\%
$$

#### Comparing Theoretical and Empirical Probabilities

To measure the difference between the theoretical and empirical probabilities, we use the absolute error:

$$
E=|\hat{P}-P|
$$

The absolute error measures the size of the difference between the two probabilities, without considering its direction.

For example, if the theoretical probability is $P=0.36$ and the empirical probability is $\hat{P}=0.354$, then:

$$
E=|0.354-0.36|=0.006
$$

This corresponds to a difference of 0.6 percentage points.

#### Why Empirical Estimates Tend to Stabilize

Simulation results vary because they depend on random outcomes. However, as the number of independent simulation repetitions increases, the empirical probability tends to approach the theoretical probability. This behavior is described by the law of large numbers (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).
    

This does not mean that the empirical probability must move closer to the theoretical probability after every additional simulation. Instead, larger numbers of repetitions generally provide more stable estimates.

#### Model Notation and Assumptions

The model uses the following notation:

- $p$: the probability of the event occurring in a single trial.
- $n$: the number of independent trials within one experiment.
- $P$: the theoretical probability of observing the event at least once within one experiment.
- $N$: the total number of independent repetitions of the experiment.
- $k$: the number of repetitions in which the event occurs at least once.
- $\hat{P}$: the empirical estimate of the theoretical probability, calculated as $k/N$.

The model relies on the following assumptions:

1. The probability $p$ remains constant across all individual trials.
2. The individual trials are independent.
3. Each simulation repetition consists of exactly $n$ trials under the same conditions.
4. The simulation repetitions are independent of one another.

Under these assumptions, the theoretical probability is $P=1-(1-p)^n$, and the empirical probability is estimated by $\hat{P}=k/N$.

The numerical examples use $p=0.2$ for simplicity. This value is illustrative and does not define how rare an event must be. The model can also be applied to smaller values of $p$.

#### Numerical Examples for H2

To illustrate the comparison between theoretical and empirical probabilities, we use the same model parameters as in the previous numerical examples: the probability of the event in a single trial is $p=0.2$, and each simulation consists of $n=2$ independent trials.

The theoretical probability of observing the event at least once is:

$$
P=1-(1-p)^n
$$

Substituting the values of $p$ and $n$ gives:

$$
P=1-(1-0.2)^2=1-0.64=0.36
$$

Therefore, the theoretical probability is $P=0.36$, or 36%.

The following examples use hypothetical simulation results to demonstrate how the empirical probability and absolute error are calculated. These values are illustrative and are not results from actual Python simulations.

**Case 1: 100 simulations ($N=100$)**

Suppose the event occurs at least once in 40 out of 100 simulations. The empirical probability is:

$$
\hat{P}=\frac{k}{N}=\frac{40}{100}=0.40=40\%
$$

The absolute error is:

$$
E=|\hat{P}-P|=|0.40-0.36|=0.04
$$

The empirical probability is 4 percentage points higher than the theoretical probability.

**Case 2: 500 simulations ($N=500$)**

Suppose the event occurs at least once in 181 out of 500 simulations. The empirical probability is:

$$
\hat{P}=\frac{181}{500}=0.362=36.2\%
$$

The absolute error is:

$$
E=|0.362-0.36|=0.002
$$

The empirical probability is 0.2 percentage points higher than the theoretical probability.

Suppose the event occurs at least once in 354 out of 1,000 simulations. The empirical probability is:

$$
E=|0.354-0.36|=0.006

**Observation**

These hypothetical examples illustrate that empirical probabilities can differ from the theoretical probability because simulation outcomes are random. The absolute error does not necessarily decrease every time the number of simulations increases.

The law of large numbers states that the relative frequency tends to approach the theoretical probability as the number of repetitions increases 
(Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)). This describes the general long-run behavior; it does not guarantee that every additional simulation will reduce the absolute error.

### 3. Mathematical Foundations for H3: Random Variation

#### Randomness in Simulation Results

A simulation uses random outcomes to approximate the behavior of a mathematical model. Even when the model parameters remain unchanged, different simulation runs can produce different results.

In this project, each simulation consists of $n$ independent trials. The event has a constant probability $p$ of occurring in each trial. The event is recorded as a success if it occurs at least once during the $n$ trials.

When the entire experiment is repeated $N$ times, the number of simulations in which the event occurs at least once can vary due to randomness. Consequently, the empirical probability

$$
\hat{P}=\frac{k}{N}
$$

may differ from the theoretical probability

$$
P=1-(1-p)^n
$$

#### Random Variation and Absolute Error

The difference between the empirical and theoretical probabilities can be measured using the absolute error:

$$
E=|\hat{P}-P|
$$

A positive absolute error indicates that the empirical estimate differs from the theoretical probability. The error can change when additional simulations are performed because the observed outcomes are random.

Importantly, increasing the number of simulations does not guarantee that the absolute error will decrease after every additional simulation. The empirical probability may move closer to the theoretical probability in one step and farther away in another.

#### Relationship to the Law of Large Numbers

The law of large numbers explains why the empirical probability tends to approach the theoretical probability as the number of independent simulation repetitions increases. However, it describes long-run behavior rather than guaranteeing a steady improvement after each repetition (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).

This distinction is the basis of H3: random variation can cause the empirical probability to fluctuate even when the underlying probability model remains unchanged.

#### Numerical Example for H3

Consider the same probability model used in the previous examples, with $p=0.2$ and $n=2$. The theoretical probability of observing the event at least once is:

$$
P=1-(1-0.2)^2=0.36
$$

Therefore, the theoretical probability remains 36% throughout the experiment.

Suppose that repeated simulations produce the following hypothetical results:

**Case 1: 100 simulations**

If the event occurs in 40 out of 100 simulations, the empirical probability is:

$$
\hat{P}_1=\frac{40}{100}=0.40=40\%
$$

The absolute error is:

$$
E_1=|0.40-0.36|=0.04
$$

**Case 2: 200 simulations**

Suppose that after 100 additional simulations, the total number of occurrences is 70 out of 200 simulations. The empirical probability becomes:

$$
\hat{P}_2=\frac{70}{200}=0.35=35\%
$$

The absolute error is:

$$
E_2=|0.35-0.36|=0.01
$$

The error has decreased from 0.04 to 0.01.

**Case 3: 300 simulations**

Suppose that after another 100 simulations, the total number of occurrences is 111 out of 300 simulations. The empirical probability is:

$$
\hat{P}_3=\frac{111}{300}=0.37=37\%
$$

The absolute error is:

$$
E_3=|0.37-0.36|=0.01
$$

**Case 4: 400 simulations**

Suppose that after another 100 simulations, the total number of occurrences is 150 out of 400 simulations. The empirical probability is:

$$
\hat{P}_4=\frac{150}{400}=0.375=37.5\%
$$

The absolute error is:

$$
E_4=|0.375-0.36|=0.015
$$

Although the number of simulations has increased, the absolute error has increased from 0.01 to 0.015.

The results are summarized below:

| Number of simulations ($N$) | Occurrences ($k$) | Empirical probability ($\hat{P}$) | Absolute error ($E$) |
|---:|---:|---:|---:|
| 100 | 40 | 40.0% | 0.040 |
| 200 | 70 | 35.0% | 0.010 |
| 300 | 111 | 37.0% | 0.010 |
| 400 | 150 | 37.5% | 0.015 |

**Observation**

These hypothetical results illustrate that the empirical probability does not necessarily move closer to the theoretical probability whenever more simulations are performed. In this example, the absolute error decreases initially but increases when the number of simulations grows from 300 to 400.

This does not contradict the law of large numbers. The law describes the long-run tendency of the empirical probability to approach the theoretical probability, not a guarantee that the absolute error will decrease at every step.
