## Mathematical Background

### 1. Mathematical Foundations for H1: Effect of the Number of Trials

#### Probability and Complementary Events

Probability measures how likely an event is to occur. Its value ranges from 0 to 1, where 0 represents an impossible event and 1 represents a certain event.

For any event A, either A occurs or its complement, denoted by Aᶜ, occurs. These two outcomes are mutually exclusive and together cover all possible outcomes.

Therefore, the probability of the complement is:

$$
P(A^c) = 1 - P(A)
$$

This relationship follows the complement rule (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).


Consider the following case: the probability of event $A$ is 0.2, or 20%.

The probability that event $A$ does not occur is:

$$
P(A^c)=1-P(A)=1-0.2=0.8
$$

Therefore, the probability of the event not occurring is 0.8, or 80%.

This illustrates that the probabilities of an event and its complement sum to 1.

#### Independent Trials

Two trials are independent when the outcome of one trial does not change the probability of an event occurring in the other trial (Illowsky & Dean, 2023, [Section 3.2](https://openstax.org/books/introductory-statistics-2e/pages/3-2-independent-and-mutually-exclusive-events)).
    

In this project, we assume that the trials are independent and that the probability of the event, denoted by $p$, remains constant across all trials. Keeping $p$ constant allows us to investigate how changing the number of trials, denoted by $n$, affects the probability of observing the event at least once.

The probability that the event does not occur in a single trial is:

$$
1-p
$$

For example, if the probability of an event in a single trial is $p = 0.2$, then the probability that it does not occur is:

$$
1-p=1-0.2=0.8
$$

Therefore, the probability of the event not occurring in a single trial is 0.8, or 80%.

#### Probability of No Occurrence in Multiple Independent Trials

If the trials are independent, the probability that the event does not occur in either of two trials is:

$$
(1-p)(1-p)=(1-p)^2
$$

For $p=0.2$, this probability is:

$$
(1-0.2)^2=0.8^2=0.64
$$

Therefore, the probability that the event does not occur in either of the two trials is 0.64, or 64%.

#### Probability of at Least One Occurrence

The probability of observing an event at least once is the complement of the probability that the event does not occur in any trial.

Using the complement rule, we obtain:

$$
P(\text{at least one occurrence})=1-P(\text{no occurrence})
$$

For $n$ independent trials, the probability of no occurrence is $(1-p)^n$. Therefore, the probability of observing the event at least once is:

$$
\boxed{P(\text{at least one occurrence})=1-(1-p)^n}
$$

In this formula, $p$ represents the probability of the event in a single trial, and $n$ represents the number of independent trials.

The multiplication rule for independent events states that the probability of all events occurring is the product of their individual probabilities (Illowsky & Dean, 2023, [Section 3.3](https://openstax.org/books/introductory-statistics-2e/pages/3-3-two-basic-rules-of-probability)).

To investigate how the number of independent trials affects the probability of observing an event at least once, we keep the probability of the event in a single trial constant at $p=0.2$, or 20%, and vary the number of trials, denoted by $n$.

Keeping $p$ constant allows us to examine the effect of the number of trials without changing the probability of the event in each individual trial.

Using the formula derived above:

$$
P(\text{at least one occurrence})=1-(1-p)^n
$$

**Case 1: One trial ($n=1$)**

$$
P=1-(1-0.2)^1=1-0.8=0.2=20\%
$$

**Case 2: Two trials ($n=2$)**

$$

$$

$$

$$

### 2. Mathematical Foundations for H2: Theoretical and Empirical Probability

#### Theoretical Probability

The theoretical probability is the probability calculated using a mathematical model. In this project, it is calculated using the formula derived above:

$$
P=1-(1-p)^n
$$

This value provides a mathematical reference against which the simulation results can be compared.

#### Empirical Probability and Relative Frequency

The empirical probability is estimated from the results of repeated simulations. It is calculated as the proportion of simulations in which the event occurs at least once.

Let $N$ denote the total number of simulations and $k$ denote the number of simulations in which the event occurs at least once. The empirical probability is:

$$
\hat{P}=\frac{k}{N}
$$

Here, $\hat{P}$ represents the empirical estimate of the theoretical probability $P$.

For example, if the event occurs at least once in 354 out of 1,000 simulations, the empirical probability is:

$$
\hat{P}=\frac{354}{1000}=0.354=35.4\%
$$

#### Comparing Theoretical and Empirical Probabilities

To measure the difference between the theoretical and empirical probabilities, we use the absolute error:

$$
E=|\hat{P}-P|
$$

The absolute error measures the size of the difference between the two probabilities, without considering its direction.

For example, if the theoretical probability is $P=0.36$ and the empirical probability is $\hat{P}=0.354$, then:

$$
E=|0.354-0.36|=0.006
$$

This corresponds to a difference of 0.6 percentage points.

#### Why Empirical Estimates Tend to Stabilize

Simulation results vary because they depend on random outcomes. However, as the number of independent simulation repetitions increases, the empirical probability tends to approach the theoretical probability. This behavior is described by the law of large numbers (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).
    

This does not mean that the empirical probability must move closer to the theoretical probability after every additional simulation. Instead, larger numbers of repetitions generally provide more stable estimates.

#### Model Notation and Assumptions

The model uses the following notation:

- $p$: the probability of the event occurring in a single trial.
- $n$: the number of independent trials within one experiment.
- $P$: the theoretical probability of observing the event at least once within one experiment.
- $N$: the total number of independent repetitions of the experiment.
- $k$: the number of repetitions in which the event occurs at least once.
- $\hat{P}$: the empirical estimate of the theoretical probability, calculated as $k/N$.

The model relies on the following assumptions:

1. The probability $p$ remains constant across all individual trials.
2. The individual trials are independent.
3. Each simulation repetition consists of exactly $n$ trials under the same conditions.
4. The simulation repetitions are independent of one another.

Under these assumptions, the theoretical probability is $P=1-(1-p)^n$, and the empirical probability is estimated by $\hat{P}=k/N$.

The numerical examples use $p=0.2$ for simplicity. This value is illustrative and does not define how rare an event must be. The model can also be applied to smaller values of $p$.

### 3. Mathematical Foundations for H3: Random Variation

#### Randomness in Simulation Results

A simulation uses random outcomes to approximate the behavior of a mathematical model. Even when the model parameters remain unchanged, different simulation runs can produce different results.

In this project, each simulation consists of $n$ independent trials. The event has a constant probability $p$ of occurring in each trial. The event is recorded as a success if it occurs at least once during the $n$ trials.

When the entire experiment is repeated $N$ times, the number of simulations in which the event occurs at least once can vary due to randomness. Consequently, the empirical probability

$$
\hat{P}=\frac{k}{N}
$$

may differ from the theoretical probability

$$
P=1-(1-p)^n
$$

#### Random Variation and Absolute Error

The difference between the empirical and theoretical probabilities can be measured using the absolute error:

$$
E=|\hat{P}-P|
$$

A positive absolute error indicates that the empirical estimate differs from the theoretical probability. The error can change when additional simulations are performed because the observed outcomes are random.

Importantly, increasing the number of simulations does not guarantee that the absolute error will decrease after every additional simulation. The empirical probability may move closer to the theoretical probability in one step and farther away in another.

#### Relationship to the Law of Large Numbers

The law of large numbers explains why the empirical probability tends to approach the theoretical probability as the number of independent simulation repetitions increases. However, it describes long-run behavior rather than guaranteeing a steady improvement after each repetition (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).

This distinction is the basis of H3: random variation can cause the empirical probability to fluctuate even when the underlying probability model remains unchanged.

## Part I — H1: Effect of the Number of Trials

### 1. Research Objective

The first part of the analysis investigates whether increasing the number of independent trials increases the theoretical probability of observing an event at least once, while keeping the single-trial event probability constant.

The analysis begins with theoretical calculations using the complement rule. It then examines multiple values of the single-trial probability to investigate whether the relationship holds across different event probabilities.

### 2. Theoretical Probability Calculations

The theoretical probability will be calculated for several values of the number of independent trials, `n`, while keeping the single-trial event probability fixed at `p = 0.2`.

The results will then be extended to additional values of `p` to examine the relationship under different model parameters.


```python
# H1: Investigate how the number of independent trials affects probability

# Define the single-trial probabilities to compare
probabilities = [0.01, 0.05, 0.1, 0.2, 0.5]  # Illustrative values of p

# Define the numbers of independent trials to investigate
trial_counts = [1, 2, 3, 5, 10, 20, 50, 80, 100, 120]  # Values of n


# Calculate theoretical probabilities for each combination of p and n
for p in probabilities:  # Repeat the analysis for each single-trial probability
    print(f"\nSingle-trial probability p = {p:.0%}") 

    for n in trial_counts:  # Examine each number of independent trials
        theoretical_probability = 1 - (1 - p) ** n

        print(f"n = {n:2}: Theoretical probability = {theoretical_probability:.2%}")
```

```python
# Visualize the theoretical probability calculations

import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter


plt.figure(figsize=(10, 6)) 


for p in probabilities:
    theoretical_probabilities = []  # Store the calculated probabilities for this p

    # Calculate the probability for each number of trials
    for n in trial_counts:
        probability = 1 - (1 - p) ** n  # Apply the complement rule
        theoretical_probabilities.append(probability)  # Save the result

    # Plot the results for the current value of p
    plt.plot(
        trial_counts,
        theoretical_probabilities,
        marker="o",
        label=f"p = {p:.0%}"
    )


plt.title("Probability of at Least One Occurrence")
plt.xlabel("Number of Independent Trials (n)")
plt.ylabel("Probability")
plt.xlim(0, 155)  
plt.xticks(trial_counts)  


plt.gca().yaxis.set_major_formatter(
    PercentFormatter(1.0)  
)


plt.legend(title="Single-trial probability")
plt.grid(True, alpha=0.3)


plt.tight_layout()
plt.show()
```

## Part II — H2: Theoretical vs. Simulated Probability

### 1. Research Objective

The second part of the analysis investigates whether empirical probabilities estimated through Monte Carlo simulation approach their corresponding theoretical probabilities as the number of simulation repetitions increases.

The experiment uses repeated independent simulations of the same probability model. Each simulation consists of n independent trials, where the probability of the event occurring in a single trial is p.



### 2. Simulation Method

For each simulation repetition, the program generates $n$ independent random outcomes. A repetition is classified as a success if the event occurs at least once among the $n$ trials.

The empirical probability is calculated as the proportion of successful simulation repetitions:

$$
\hat{P} = \frac{K}{N}
$$

where $K$ is the number of successful repetitions and $N$ is the total number of simulation repetitions.

The empirical probability is then compared with the theoretical probability:

$$
P = 1 - (1-p)^n
$$

The absolute error is calculated as:

$$
\text{Absolute Error} = |\hat{P} - P|
$$


```python
import random

p = 0.2
n = 10

simulation_counts = [50, 100, 500, 1000, 5000, 10000, 20000, 50000, 100000, 200000]

theoretical_probability = 1 - (1 - p) ** n

# Store the results for visualization
empirical_probabilities = []
absolute_errors = []

print(f"Theoretical probability: {theoretical_probability:.2%}")

# Run the simulation for each selected number of repetitions
for N in simulation_counts:
    successful_repetitions = 0

    # Repeat the experiment N times
    for _ in range(N):
        # Generate n independent outcomes for one simulation
        outcomes = [random.random() < p for _ in range(n)]

        # Check whether the event occurred at least once
        if any(outcomes):
            successful_repetitions += 1

    # Calculate the empirical probability and absolute error
    empirical_probability = successful_repetitions / N
    absolute_error = abs(empirical_probability - theoretical_probability)

    # Save the results for visualization
    empirical_probabilities.append(empirical_probability)
    absolute_errors.append(absolute_error)

    
    print(f"\nNumber of simulations: {N}")
    print(f"Successful repetitions: {successful_repetitions}")
    print(f"Empirical probability: {empirical_probability:.2%}")
    print(f"Absolute error: {absolute_error:.4f}")
```

### 3. Simulation Results

The simulation results are used to compare the empirical probability with the theoretical probability for different numbers of simulation repetitions.

The absolute error measures the difference between the empirical and theoretical probabilities. The results are examined to assess how the empirical estimate behaves as the number of simulation repetitions increases.


### 4. Visualization of Simulation Results

The following visualization compares the empirical probabilities obtained from Monte Carlo simulations with the theoretical probability. It illustrates how the empirical estimates vary with the number of simulation repetitions.


```python
# Visualize how the empirical probability compares with the theoretical probability

import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

plt.figure(figsize=(10, 6))

plt.plot(
    simulation_counts,
    empirical_probabilities,
    marker="o",
    label="Empirical probability"
)

plt.axhline(
    y=theoretical_probability,
    linestyle="--",
    label="Theoretical probability"
)

plt.xscale("log")
plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))

plt.title("Theoretical vs. Empirical Probability")
plt.xlabel("Number of Simulation Repetitions (N, logarithmic scale)")
plt.ylabel("Probability")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```


### 5. Results and Interpretation

The simulation results illustrate how the empirical probability compares with the theoretical probability as the number of simulation repetitions increases.

For a small number of repetitions, the empirical estimates may show larger deviations from the theoretical probability. As the number of repetitions increases, the estimates generally become more stable and tend to be closer to the theoretical value of approximately 89.26%.

However, this relationship is not perfectly smooth or monotonic. Each simulation count is evaluated using a separate simulation run, so random variation may cause an estimate based on more repetitions to be farther from the theoretical probability than an estimate based on fewer repetitions.


Overall, the results are consistent with the expected behavior described by Hypothesis H2. As the number of simulation repetitions increases, the empirical probability estimates tend to remain closer to the theoretical probability, although random variation may cause individual estimates to deviate from this trend.




## Part III — H3: Random Variation Across Simulation Runs

### 2. Experiment Design

The experiment uses the same probability model and fixed parameters as in the previous section:

- Number of independent trials per simulation: $n = 10$
- Number of repetitions per simulation run: $N = 1000$

import random

run_counts = [10, 30, 100]

theoretical_probability = 1 - (1 - p) ** n

# Store results separately for each number of simulation runs
h3_results = {}

for R in run_counts:
    empirical_probabilities = []
    absolute_errors = []

    # Run the experiment independently R times
    for _ in range(R):
        successful_repetitions = 0

        for _ in range(N):
            outcomes = [random.random() < p for _ in range(n)]

            if any(outcomes):
                successful_repetitions += 1

        empirical_probability = successful_repetitions / N
        absolute_error = abs(
            empirical_probability - theoretical_probability
        )

        empirical_probabilities.append(empirical_probability)
        absolute_errors.append(absolute_error)

    h3_results[R] = {
        "probabilities": empirical_probabilities,
        "errors": absolute_errors
    }

    print(f"\nNumber of independent runs: {R}")
    print(f"Mean empirical probability: "
          f"{sum(empirical_probabilities) / R:.2%}")
    print(f"Mean absolute error: "
          f"{sum(absolute_errors) / R:.4f}")

print(f"\nTheoretical probability: {theoretical_probability:.2%}")

Each point represents the empirical probability obtained from one independent simulation run. The red dashed horizontal line indicates the theoretical probability, calculated as:

P = 1 - (1-p)^n
$$

The visualization illustrates how empirical estimates vary around the theoretical probability while the model parameters remain constant.

### Results and Interpretation

The experiment investigates how empirical probability varies across independent simulation runs when the model parameters remain constant. The single-trial event probability is set to \(p = 0.2\), the number of trials per experiment is \(n = 10\), and each independent run consists of \(N = 1000\) simulation repetitions.

The theoretical probability of observing at least one occurrence is:

$$
P = 1-(1-p)^n = 1-(1-0.2)^{10} \approx 89.26\%
$$

#### 1. Results for \(R = 10\)

With 10 independent simulation runs, the empirical probabilities vary around the theoretical probability. Some estimates are above the theoretical value, while others are below it.

Because the number of observations is relatively small, the results provide only a limited view of the possible random variation. Nevertheless, the differences between successive estimates illustrate that empirical probability does not necessarily move closer to the theoretical value after every additional run.

#### 2. Results for \(R = 30\)

Increasing the number of independent runs to 30 provides more observations of the variability in empirical probability.

The estimates continue to fluctuate around the theoretical value. An estimate that is relatively close to the theoretical probability may be followed by one that is farther away. This behavior is consistent with the random nature of the simulation process.

The additional runs provide a broader view of the observed variation, but they do not eliminate the differences between individual estimates.

#### 3. Results for \(R = 100\)

With 100 independent simulation runs, the experiment produces a larger collection of empirical probability estimates.

The graph shows that the estimates continue to vary, with observations both above and below the theoretical probability. The larger number of runs provides more information about the distribution of these estimates.

Increasing the number of independent runs does not necessarily make each individual estimate more accurate. Instead, it allows the variation across independent estimates to be examined more thoroughly.

#### 4. Evaluation of Hypothesis H3

Hypothesis H3 states that individual simulation results vary due to randomness and that empirical probability does not necessarily become closer to the theoretical probability after every additional simulation run.

The observed results are consistent with this hypothesis. Across the three settings, the empirical probabilities fluctuate around the theoretical value rather than approaching it in a strictly monotonic manner.

The experiment therefore illustrates the distinction between random variation in individual estimates and the overall behavior of a collection of estimates. The results support H3 within the conditions of the simulation, although the visualization alone does not constitute a formal proof.

#### 5. Limitations

The experiment uses a fixed probability of \(p = 0.2\), a fixed number of trials per experiment of \(n = 10\), and a fixed number of repetitions per run of \(N = 1000\). Consequently, the findings illustrate random variation under these particular conditions and should not be interpreted as a complete analysis of all possible parameter settings.

Furthermore, increasing \(R\) increases the number of independent estimates available for analysis, but it does not directly improve the accuracy of each individual estimate. The variability of an individual estimate depends primarily on the number of repetitions \(N\), as well as on the underlying probability model.

#### 6. Conclusion

The simulation demonstrates that empirical probability varies across independent runs, even when the model parameters remain unchanged. The estimates fluctuate around the theoretical probability of approximately 89.26%, and individual results do not necessarily become closer to this value with each successive run.

Comparing \(R = 10\), \(R = 30\), and \(R = 100\) provides progressively more observations of this variation. Overall, the findings are consistent with Hypothesis H3 and illustrate the role of randomness in Monte Carlo simulation.
