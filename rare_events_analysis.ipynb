## Research Hypotheses

**H1 - Effect of the Number of Trials**

If the probability of a rare event in each independent trial remains constant and satisfies $0<p<1$, then increasing the number of trials will increase the probability of observing the event at least once.

**H2 - Theoretical and simulated probabilities**

## Mathematical Background

### 1. Mathematical Foundations for H1: Effect of the Number of Trials

#### Probability and Complementary Events

Probability measures how likely an event is to occur. Its value ranges from **0** to **1**, where **0** represents an impossible event and **1** represents a certain event.

For any event $A$, either $A$ occurs or its complement, denoted by $Aᶜ$, occurs. These two outcomes are mutually exclusive and together cover all possible outcomes.

Therefore, the probability of the complement is:

$$
P(A^c) = 1 - P(A)
$$

This relationship follows the complement rule (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).


Consider the following case: the probability of event $A$ is **0.2**, or **20%**.

The probability that event $A$ does not occur is:

$$
P(A^c)=1-P(A)=1-0.2=0.8
$$

Therefore, the probability of the event not occurring is **0.8**, or **80%**.

This illustrates that the probabilities of an event and its complement sum to **1**.

#### Independent Trials

Two trials are independent when the outcome of one trial does not change the probability of an event occurring in the other trial (Illowsky & Dean, 2023, [Section 3.2](https://openstax.org/books/introductory-statistics-2e/pages/3-2-independent-and-mutually-exclusive-events)).

Independence is important because it allows the probability of no occurrence across all $n$ trials to be calculated as the product of the probabilities of no occurrence in each trial. If the probability of the event in each trial is $p$, the probability of no occurrence in one trial is $1-p$. Under independence, the probability of no occurrence in all $n$ trials is therefore $(1-p)^n$. If the trials are dependent, this product generally cannot be used without additional information about how the outcomes are related.
    

In this project, we assume that the trials are independent and that the probability of the event, denoted by $p$, remains constant across all trials. Keeping $p$ constant allows us to investigate how changing the number of trials, denoted by $n$, affects the probability of observing the event at least once.

The probability that the event does not occur in a single trial is:

$$
1-p
$$

For example, if the probability of an event in a single trial is $p = 0.2$, then the probability that it does not occur is:

$$
1-p=1-0.2=0.8
$$

Therefore, the probability of the event not occurring in a single trial is **0.8**, or **80%**.

#### Probability of No Occurrence in Multiple Independent Trials

If the trials are independent, the probability that the event does not occur in either of two trials is:

$$
(1-p)(1-p)=(1-p)^2
$$

For $p=0.2$, this probability is:

$$
(1-0.2)^2=0.8^2=0.64
$$

Therefore, the probability that the event does not occur in either of the two trials is **0.64**, or **64%**.

#### Probability of at Least One Occurrence

The probability of observing an event at least once is the complement of the probability that the event does not occur in any trial.

Using the complement rule, we obtain:

$$
P(\text{at least one occurrence})=1-P(\text{no occurrence})
$$

For $n$ independent trials, the probability of no occurrence is $(1-p)^n$. Therefore, the probability of observing the event at least once is:

$$
\boxed{P(\text{at least one occurrence})=1-(1-p)^n}
$$

In this formula, $p$ represents the probability of the event in a single trial, and $n$ represents the number of independent trials.

The multiplication rule for independent events states that the probability of all events occurring is the product of their individual probabilities (Illowsky & Dean, 2023, [Section 3.3](https://openstax.org/books/introductory-statistics-2e/pages/3-3-two-basic-rules-of-probability)).


#### Numerical Examples for H1

To investigate how the number of independent trials affects the probability of observing an event at least once, we keep the probability of the event in a single trial constant at $p=0.2$, or **20%**, and vary the number of trials, denoted by $n$.

Keeping $p$ constant allows us to examine the effect of the number of trials without changing the probability of the event in each individual trial.

Using the formula derived above:

$$
P(\text{at least one occurrence})=1-(1-p)^n
$$

**Case 1: One trial ($n=1$)**

$$
P=1-(1-0.2)^1=1-0.8=0.2=20\%
$$

**Case 2: Two trials ($n=2$)**

$$
P=1-(1-0.2)^2=1-0.64=0.36=36\%
$$

**Case 3: Three trials ($n=3$)**

$$
P=1-(1-0.2)^3=1-0.512=0.488=48.8\%
$$

**Case 4: Five trials ($n=5$)**

$$
P=1-(1-0.2)^5=1-0.32768=0.67232=67.232\%
$$

**Case 5: Ten trials ($n=10$)**

$$
P=1-(1-0.2)^{10}=1-0.1073741824=0.8926258176\approx89.26\%
$$

The results can be summarized as follows:

| Number of trials ($n$) | Probability of at least one occurrence |
|---:|---:|
| 1 | 20.00% |
| 2 | 36.00% |
| 3 | 48.80% |

**Observation**

The theoretical probability of observing the event at least once increases as the number of independent trials increases, even though the probability in each individual trial remains constant at **20%**.

The increase is not linear: adding more trials does not produce the same increase in percentage points each time. The probability approaches **100%** as the number of trials increases, provided that $p>0$ remains constant and the trials are independent.


### 2. Mathematical Foundations for H2: Theoretical and Empirical Probability

#### Theoretical Probability

The theoretical probability is the probability calculated using a mathematical model. In this project, it is calculated using the formula derived above:

$$
P=1-(1-p)^n
$$

This value provides a mathematical reference against which the simulation results can be compared.

#### Empirical Probability and Relative Frequency

The empirical probability is estimated from the results of repeated simulations. It is calculated as the proportion of simulations in which the event occurs at least once.

Let $N$ denote the total number of simulations and $k$ denote the number of simulations in which the event occurs at least once. The empirical probability is:

$$
\hat{P}=\frac{k}{N}
$$

Here, $\hat{P}$ represents the empirical estimate of the theoretical probability $P$.

For example, if the event occurs at least once in **354** out of **1,000** simulations, the empirical probability is:

$$
\hat{P}=\frac{354}{1000}=0.354=35.4\%
$$

#### Comparing Theoretical and Empirical Probabilities

To measure the difference between the theoretical and empirical probabilities, we use the absolute error:

$$
E=|\hat{P}-P|
$$

The absolute error measures the size of the difference between the two probabilities, without considering its direction.

For example, if the theoretical probability is $P=0.36$ and the empirical probability is $\hat{P}=0.354$, then:

$$
E=|0.354-0.36|=0.006
$$

This corresponds to a difference of **0.6** percentage points.

#### Why Empirical Estimates Tend to Stabilize

Simulation results vary because they depend on random outcomes. However, as the number of independent simulation repetitions increases, the empirical probability tends to approach the theoretical probability. This behavior is described by the law of large numbers (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).
    

This does not mean that the empirical probability must move closer to the theoretical probability after every additional simulation. Instead, larger numbers of repetitions generally provide more stable estimates.

#### Model Notation and Assumptions

The model uses the following notation:

- $p$: the probability of the event occurring in a single trial.
- $n$: the number of independent trials within one experiment.
- $P$: the theoretical probability of observing the event at least once within one experiment.
- $N$: the total number of independent repetitions of the experiment.
- $k$: the number of repetitions in which the event occurs at least once.
- $\hat{P}$: the empirical estimate of the theoretical probability, calculated as $k/N$.

The model relies on the following assumptions:

-  The probability $p$ remains constant across all individual trials.
-  The individual trials are independent.
-  Each simulation repetition consists of exactly $n$ trials under the same conditions.
-  The simulation repetitions are independent of one another.

Under these assumptions, the theoretical probability is $P=1-(1-p)^n$, and the empirical probability is estimated by $\hat{P}=k/N$.

The numerical examples use $p=0.2$ for simplicity. This value is illustrative and does not define how rare an event must be. The model can also be applied to smaller values of $p$.

### 3. Mathematical Foundations for H3: Random Variation

#### Randomness in Simulation Results

A simulation uses random outcomes to approximate the behavior of a mathematical model. Even when the model parameters remain unchanged, different simulation runs can produce different results.

In this project, each simulation consists of $n$ independent trials. The event has a constant probability $p$ of occurring in each trial. The event is recorded as a success if it occurs at least once during the $n$ trials.

When the entire experiment is repeated $N$ times, the number of simulations in which the event occurs at least once can vary due to randomness. Consequently, the empirical probability

$$
\hat{P}=\frac{k}{N}
$$

may differ from the theoretical probability

$$
P=1-(1-p)^n
$$

#### Random Variation and Absolute Error

The difference between the empirical and theoretical probabilities can be measured using the absolute error:

$$
E=|\hat{P}-P|
$$

A positive absolute error indicates that the empirical estimate differs from the theoretical probability. The error can change when additional simulations are performed because the observed outcomes are random.

Importantly, increasing the number of simulations does not guarantee that the absolute error will decrease after every additional simulation. The empirical probability may move closer to the theoretical probability in one step and farther away in another.

#### Relationship to the Law of Large Numbers

The law of large numbers explains why the empirical probability tends to approach the theoretical probability as the number of independent simulation repetitions increases. However, it describes long-run behavior rather than guaranteeing a steady improvement after each repetition (Illowsky & Dean, 2023, [Section 3.1](https://openstax.org/books/introductory-statistics-2e/pages/3-1-terminology)).

This distinction is the basis of H3: random variation can cause the empirical probability to fluctuate even when the underlying probability model remains unchanged.

### Numerical Example for H3

$$

$$

Each independent simulation run consists of $N = 1000$ repetitions. Let $k$ denote the number of repetitions in which the event occurs at least once. The empirical probability for one simulation run is:

$$
\hat{P} = \frac{k}{N}
$$

For example, if the event occurs in $k = 895$ out of $N = 1000$ repetitions, the empirical probability is:

$$
\hat{P} = \frac{895}{1000} = 0.895 = 89.5\%
$$

This estimate differs from the theoretical probability of approximately **89.26%**. Such a difference is expected because the simulation is based on random outcomes.

The experiment repeats this process independently for $R = 10$, $R = 30$, and $R = 100$ simulation runs. Increasing $R$ provides more independent empirical probability estimates for examining random variation. It does not guarantee that each estimate becomes closer to the theoretical probability or that the observed error decreases monotonically.

# Python Simulation and Experimental Analysis

## Part I - H1: Effect of the Number of Trials

### 1. Research Objective

The first part of the analysis investigates whether increasing the number of independent trials increases the theoretical probability of observing an event at least once, while keeping the single-trial event probability constant.

The analysis begins with theoretical calculations using the complement rule. It then examines multiple values of the single-trial probability to investigate whether the relationship holds across different event probabilities.

### 2. Theoretical Probability Calculations

The theoretical probability will be calculated for several values of the number of independent trials, $n$, while keeping the single-trial event probability fixed at $p = 0.2$.

The results will then be extended to additional values of $p$ to examine the relationship under different model parameters.


```python
# H1: Investigate how the number of independent trials affects probability

# Define the single-trial probabilities to compare
probabilities = [0.01, 0.05, 0.1, 0.2, 0.5]  # Illustrative values of p

# Define the numbers of independent trials to investigate
trial_counts = [1, 2, 3, 5, 10, 20, 50, 80, 100, 120]  # Values of n


# Calculate theoretical probabilities for each combination of p and n
for p in probabilities:  # Repeat the analysis for each single-trial probability
    print(f"\nSingle-trial probability p = {p:.0%}") 

    for n in trial_counts:  # Examine each number of independent trials
        theoretical_probability = 1 - (1 - p) ** n

        print(f"n = {n:2}: Theoretical probability = {theoretical_probability:.2%}")
```

```python
# Visualize the theoretical probability calculations

import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter


plt.figure(figsize=(10, 6)) 


for p in probabilities:
    theoretical_probabilities = []  # Store the calculated probabilities for this p

    # Calculate the probability for each number of trials
    for n in trial_counts:
        probability = 1 - (1 - p) ** n  # Apply the complement rule
        theoretical_probabilities.append(probability)  # Save the result

    # Plot the results for the current value of p
    plt.plot(
        trial_counts,
        theoretical_probabilities,
        marker="o",
        label=f"p = {p:.0%}"
    )


plt.title("Probability of at Least One Occurrence")
plt.xlabel("Number of Independent Trials (n)")
plt.ylabel("Probability")
plt.xlim(0, 155)  
plt.xticks(trial_counts)  


plt.gca().yaxis.set_major_formatter(
    PercentFormatter(1.0)  
)


plt.legend(title="Single-trial probability")
plt.grid(True, alpha=0.3)


plt.tight_layout()
plt.show()
```

### 3. Results and Interpretation

The experiment investigates how the theoretical probability of observing an event at least once changes as the number of independent trials increases. The analysis considers several single-trial probabilities, $p = 0.01$, $p = 0.05$, $p = 0.1$, $p = 0.2$, and $p = 0.5$, while evaluating the probability for different values of $n$.

The rate of increase depends on the single-trial probability. For example, when $p = 0.5$, the probability approaches **100%** much faster than when $p = 0.01$. This is because the probability of no occurrence in a single trial is **0.5** when $p = 0.5$, compared with **0.99** when $p = 0.01$. Across independent trials, these probabilities are raised to the power $n$. Consequently, the probability of no occurrence decreases more rapidly when the single-trial event probability $p$ is higher.
The results are consistent with the formula for the
probability of at least one occurrence:

$$
P(\text{at least one}) = 1 - (1-p)^n
$$

The calculations and graph illustrate this relationship,
while the mathematical argument presented earlier proves
that the probability increases strictly with $n$
whenever $0 < p < 1$.


For large values of $n$, some calculated probabilities are displayed as 100.00%. However, these values are rounded to two decimal places and are not necessarily exactly equal to 100%. For example, when $p = 0.2$ and $n = 50$, the theoretical probability is approximately **99.9986%**, which is displayed as **100.00%**. For any fixed $0 < p < 1$ and finite $n \geq 1$, the probability remains strictly below **100%**, because the probability of no occurrence, $(1-p)^n$, remains positive. As $n$ increases, this probability of no occurrence approaches zero, so the probability of at least one occurrence approaches 100%.

This limiting result assumes that the single-trial probability remains fixed at a value $0 < p < 1$. If $p = 0$, the probability of at least one occurrence remains $0$, while if $p = 1$, it is already $1$ for every $n \geq 1\$.



## Part II - H2: Theoretical vs. Simulated Probability

### 1. Research Objective

The second part of the analysis investigates whether empirical probabilities estimated through Monte Carlo simulation approach their corresponding theoretical probabilities as the number of simulation repetitions increases.

The experiment uses repeated independent simulations of the same probability model. Each simulation consists of $n$ independent trials, where the probability of the event occurring in a single trial is $p$.



### 2. Simulation Method

For each simulation repetition, the program generates $n$ independent random outcomes. A repetition is classified as a success if the event occurs at least once among the $n$ trials.

The empirical probability is calculated as the proportion of successful simulation repetitions:

$$
\hat{P} = \frac{k}{N}
$$

where $k$ is the number of successful repetitions and $N$ is the total number of simulation repetitions.

The empirical probability is then compared with the theoretical probability:

$$
P = 1 - (1-p)^n
$$

The absolute error is calculated as:

$$
\text{Absolute Error} = |\hat{P} - P|
$$

The absolute error measures the magnitude of the difference between an empirical estimate and the theoretical probability, but it does not indicate whether the estimate is above or below the theoretical value. It also describes the error of a particular estimate rather than guaranteeing how the error will change in a later simulation. Therefore, the absolute errors should be interpreted across different values of $N$ to examine the overall behavior of the estimates, rather than to expect a steady decrease at every step.


```python
import random

# Set a fixed random seed to make the simulation results reproducible.
# This initializes the pseudo-random number generator to a known state.
# Re-running the same code with the same sequence of random operations
# produces the same results.
# The value 42 is an arbitrary choice with no special mathematical meaning.
random.seed(42)

p = 0.2
n = 10

simulation_counts = [50, 100, 500, 1000, 5000, 10000, 20000, 50000, 100000, 200000]

theoretical_probability = 1 - (1 - p) ** n

# Store the results for visualization
empirical_probabilities = []
absolute_errors = []

print(f"Theoretical probability: {theoretical_probability:.2%}")

# Run the simulation for each selected number of repetitions
for N in simulation_counts:
    successful_repetitions = 0

    # Repeat the experiment N times
    for _ in range(N):
        # Generate n independent outcomes for one simulation
        outcomes = [random.random() < p for _ in range(n)]

        # Check whether the event occurred at least once
        if any(outcomes):
            successful_repetitions += 1

    # Calculate the empirical probability and absolute error
    empirical_probability = successful_repetitions / N
    absolute_error = abs(empirical_probability - theoretical_probability)

    # Save the results for visualization
    empirical_probabilities.append(empirical_probability)
    absolute_errors.append(absolute_error)

    
    print(f"\nNumber of simulations: {N}")
    print(f"Successful repetitions: {successful_repetitions}")
    print(f"Empirical probability: {empirical_probability:.2%}")
    print(f"Absolute error: {absolute_error:.6f}")
```

### 3. Simulation Results

The simulation results are used to compare the empirical probability with the theoretical probability for different numbers of simulation repetitions.

The absolute error measures the difference between the empirical and theoretical probabilities. The results are examined to assess how the empirical estimate behaves as the number of simulation repetitions increases.


### 4. Visualization of Simulation Results

The following visualization compares the empirical probabilities obtained from Monte Carlo simulations with the theoretical probability. It illustrates how the empirical estimates vary with the number of simulation repetitions.


```python
# Visualize how the empirical probability compares with the theoretical probability

import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

plt.figure(figsize=(10, 6))

plt.plot(
    simulation_counts,
    empirical_probabilities,
    marker="o",
    label="Empirical probability"
)

plt.axhline(
    y=theoretical_probability,
    linestyle="--",
    label="Theoretical probability"
)

plt.xscale("log")
plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))

plt.title("Theoretical vs. Empirical Probability")
plt.xlabel("Number of Simulation Repetitions (N, logarithmic scale)")
plt.ylabel("Probability")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```


### 5. Results and Interpretation

The simulation results illustrate how the empirical probability compares with the theoretical probability as the number of simulation repetitions increases.

For a small number of repetitions, the empirical estimates may show larger deviations from the theoretical probability. As the number of repetitions increases, the estimates generally become more stable and tend to be closer to the theoretical value of approximately **89.26%**.

However, this relationship is not perfectly smooth or monotonic. Each simulation count is evaluated using a separate simulation run, so random variation may cause an estimate based on more repetitions to be farther from the theoretical probability than an estimate based on fewer repetitions.


Overall, the simulation results are consistent with Hypothesis H2. As the number of simulation repetitions increases, the empirical probability is expected to converge toward the theoretical probability under the assumptions of the model. The observed results illustrate this general tendency, although the absolute error does not decrease monotonically for every increase in the number of repetitions. The findings are consistent with the law of large numbers, which describes long-run convergence rather than guaranteed improvement at every individual step.




## Part III - H3: Random Variation Across Simulation Runs

### 1. Research Objective

The third part of the analysis investigates how random variation affects empirical probability estimates across independent simulation runs.

Even when the model parameters and the number of simulation repetitions remain constant, different runs may produce different empirical probabilities.

This experiment examines how empirical probability estimates and their absolute errors vary across independent simulation runs when the model parameters remain constant.

### 2. Experiment Design

The experiment uses the same probability model and the same values of the single-trial probability and the number of trials per experiment as in H2.

The model parameters are kept constant throughout the experiment:

- Single-trial event probability: $p = 0.2$
- Number of independent trials per experiment: $n = 10$
- Number of repetitions per simulation run: $N = 1000$
- Number of independent simulation runs: $R = 10$, $R = 30$, $R = 100$

Each simulation run consists of $N = 1000$ repetitions of the same experiment. For each run, the empirical probability and its absolute error relative to the theoretical probability are calculated.

The experiment compares three settings with different numbers of independent simulation runs: $R = 10$, $R = 30$, and $R = 100$. This allows the variability of individual empirical probability estimates to be examined using progressively larger collections of simulation results.


```python
import random

random.seed(42)

p = 0.2
n = 10
N = 1000
run_counts = [10, 30, 100]

theoretical_probability = 1 - (1 - p) ** n

# Store results separately for each number of simulation runs
h3_results = {}

for R in run_counts:
    empirical_probabilities = []
    absolute_errors = []

    # Run the experiment independently R times
    for _ in range(R):
        successful_repetitions = 0

        for _ in range(N):
            outcomes = [random.random() < p for _ in range(n)]

            if any(outcomes):
                successful_repetitions += 1

        empirical_probability = successful_repetitions / N
        absolute_error = abs(
            empirical_probability - theoretical_probability
        )

        empirical_probabilities.append(empirical_probability)
        absolute_errors.append(absolute_error)

    h3_results[R] = {
        "probabilities": empirical_probabilities,
        "errors": absolute_errors
    }

    print(f"\nNumber of independent runs: {R}")
    print(f"Mean empirical probability: "
          f"{sum(empirical_probabilities) / R:.2%}")
    print(f"Mean absolute error: "
          f"{sum(absolute_errors) / R:.4f}")

print(f"\nTheoretical probability: {theoretical_probability:.2%}")
```

### 3. Visualization of Random Variation

The following visualization compares empirical probability estimates obtained from **10**, **30**, and **100** independent simulation runs.

Each point represents the empirical probability obtained from one independent simulation run consisting of $N = 1000$ repetitions. The red dashed horizontal line represents the theoretical probability:

$$
P = 1-(1-p)^n
$$

The visualization illustrates the random variation of empirical estimates around the theoretical probability while the model parameters remain constant. Comparing the three settings also shows how increasing the number of independent runs provides more observations of this variation.


```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

# Compare individual empirical probabilities across independent runs
fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)

for ax, R in zip(axes, run_counts):
    probabilities = h3_results[R]["probabilities"]
    run_numbers = list(range(1, R + 1))

    ax.scatter(
        run_numbers,
        probabilities,
        s=35,
        alpha=0.8,
        label="Empirical probability"
    )

    ax.axhline(
        y=theoretical_probability,
        linestyle="--",
        color="red",
        linewidth=1.5,
        label="Theoretical probability"
    )

    ax.set_title(f"{R} Independent Runs")
    ax.set_xlabel("Simulation Run")
    ax.yaxis.set_major_formatter(PercentFormatter(1.0))
    ax.grid(True, alpha=0.25)

axes[0].set_ylabel("Empirical Probability")
axes[0].legend(fontsize=8, loc="lower left")

fig.suptitle(
    "Random Variation in Empirical Probability",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

### 4. Quantitative Analysis of Random Variation

The empirical probability estimates are further examined using summary statistics. For each number of independent simulation runs, the mean empirical probability, standard deviation, mean absolute error, and maximum absolute error are calculated.

These statistics provide a quantitative summary of the variation observed across independent simulation runs.

The number of repetitions per run, $N$, and the number of independent runs, $R$, serve different purposes. Increasing $N$ generally reduces the sampling variability of an individual empirical probability estimate under the model assumptions. Increasing $R$, while keeping $N$ fixed, provides more estimates with which to examine the distribution and variability of those estimates. However, summary statistics calculated from a finite collection of runs can fluctuate as $R$ changes, so an observed standard deviation, mean absolute error, or maximum absolute error is not guaranteed to decrease monotonically.

For a fixed number of repetitions per run, the theoretical standard deviation of an empirical probability estimate can be expressed as:

$$
\sigma_{\hat{P}}=\sqrt{\frac{P(1-P)}{N}}
$$

For $P \approx 0.8926$ and $N = 1000$, this gives a theoretical standard deviation of approximately **0.0098**. The observed standard deviations across the independent simulation runs are expected to vary around this value because they are estimated from finite samples.

```python
import statistics

# Summarize the variation across independent simulation runs

for R in run_counts:
    probabilities = h3_results[R]["probabilities"]
    errors = h3_results[R]["errors"]

    mean_probability = statistics.mean(probabilities)
    standard_deviation = statistics.stdev(probabilities)
    mean_absolute_error = statistics.mean(errors)
    maximum_absolute_error = max(errors)

    print(f"\nNumber of independent runs: {R}")
    print(f"Mean empirical probability: {mean_probability:.2%}")
    print(f"Standard deviation: {standard_deviation:.4f}")
    print(f"Mean absolute error: {mean_absolute_error:.4f}")
    print(f"Maximum absolute error: {maximum_absolute_error:.4f}")
```

### 5. Visualization of Absolute Error

The following visualization compares the absolute errors of empirical probability estimates across independent simulation runs. Each point represents the absolute difference between an empirical probability estimate and the theoretical probability.

The visualization complements the comparison of empirical probabilities by showing the magnitude of the estimation errors directly.

```python
# Visualize absolute errors across independent simulation runs

fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)

for ax, R in zip(axes, run_counts):
    errors = h3_results[R]["errors"]
    run_numbers = list(range(1, R + 1))

    ax.scatter(
        run_numbers,
        errors,
        s=35,
        alpha=0.8,
        label="Absolute error"
    )

    ax.set_title(f"{R} Independent Runs")
    ax.set_xlabel("Simulation Run")
    ax.grid(True, alpha=0.25)

axes[0].set_ylabel("Absolute Error")

fig.suptitle(
    "Absolute Error Across Independent Simulation Runs",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

### 6. Results and Interpretation

The experiment investigates how empirical probability varies across independent simulation runs when the model parameters remain constant. The single-trial event probability is set to $p = 0.2$, the number of trials per experiment is $n = 10$, and each independent run consists of $N = 1000$ simulation repetitions.

The theoretical probability of observing at least one occurrence is:

$$
P = 1-(1-p)^n = 1-(1-0.2)^{10} \approx 89.26\%
$$

#### 1. Results for \(R = 10\)

For **10** independent simulation runs, the mean empirical probability is **89.74%**, with a standard deviation of **0.0120**. The mean absolute error is **0.0103** and the maximum absolute error is **0.0236**.

The mean empirical probability is slightly higher than the theoretical probability. The estimates vary across runs, illustrating that independent simulations can produce different results even when the model parameters remain unchanged.

#### 2. Results for \(R = 30\)

For **30** independent simulation runs, the mean empirical probability is **89.19%**, with a standard deviation of **0.0086**. The mean absolute error is **0.0066** and the maximum absolute error is **0.0224**.

The mean empirical probability remains relatively close to the theoretical probability of approximately **89.26%**. The mean absolute error is lower than the value observed for $R = 10$, while the standard deviation is also lower. However, these differences are observed in the current simulation results and do not establish a general rule that these statistics must decrease as the number of independent runs increases.

#### 3. Results for \(R = 100\)

For **100** independent simulation runs, the mean empirical probability is **89.15%**, with a standard deviation of **0.0094**. The mean absolute error is **0.0075** and the maximum absolute error is **0.0236**.

The larger collection of runs provides more observations of the variation between independent empirical estimates. The mean empirical probability remains close to the theoretical value, while individual estimates continue to fluctuate around it.

The standard deviation and error statistics do not decrease monotonically as $R$ increases. This is expected because the three settings use separate collections of independently generated simulation runs.

#### 4. Evaluation of Hypothesis H3

The results are consistent with Hypothesis H3. The mean empirical probabilities for $R = 10$, $R = 30$, and $R = 100$ are **89.74%**, **89.19%**, and **89.15%**, respectively, compared with the theoretical probability of approximately **89.26%**.

The observed standard deviations and error measures vary across the three settings and do not decrease monotonically as the number of independent runs increases. This is consistent with the random variation expected in finite simulation samples. The results illustrate H3 under the conditions examined, but they do not constitute a formal proof of the hypothesis.
#### 5. Limitations

The experiment uses a fixed single-trial probability of $p = 0.2$, a fixed number of trials per experiment of $n = 10$, and a fixed number of repetitions per run of $N = 1000$. Consequently, the findings describe random variation under these particular model conditions and should not be interpreted as a complete analysis of all possible parameter settings.

The values of $R$ determine how many independent empirical probability estimates are collected for analysis. Increasing $R$ provides more observations of the variation between estimates, but it does not directly improve the accuracy of each individual estimate. The variability of an individual estimate depends primarily on $N$ and the underlying probability model.

Furthermore, the three settings use separate collections of independent simulation runs. They are not nested experiments in which the results from $R = 10$ are retained and then extended to $R = 30$ and $R = 100$. Therefore, differences between their summary statistics should be interpreted as observed variation between samples, not as a guaranteed trend caused by increasing $R$.
